```python
import pickle
import numpy as np
from scipy.stats import ttest_rel
import pingouin as pg
import pandas as pd
import pingouin as pg
import os
import matplotlib.pyplot as plt
from glob import glob
import mne
```

```python
bids_root = "/projects/p52964/EAM1/data-bids/"
deriv_dir = os.path.join(bids_root, 'derivatives', 'epochs-ffr')

rater1 = pickle.load(open(os.path.join(deriv_dir, 'onset_times_kevin.pkl'), 'rb'))
rater2 = pickle.load(open(os.path.join(deriv_dir, 'onset_times_laura_3.pkl'), 'rb'))
rater3 = pickle.load(open(os.path.join(deriv_dir, 'onset_times_tiebreaker_2.pkl'), 'rb'))
concensus = pickle.load(open(os.path.join(deriv_dir, 'onset_times_34.pkl'), 'rb'))
print()
```

```python
active_subs = [5, 17, 19, 21, 31, 34]
passive_subs = [14, 17, 21, 25, 31]

# read in epochs
act_pos_avgs = []
act_neg_avgs = []
act_eps = [sorted(glob(deriv_dir+f'/sub-{sub_num:02d}_task-active_run-all_event-stimtrack_epochs.fif') 
                for sub_num in active_subs)]
for act_ep in act_eps[0]:
    try:
        sub_active_epochs = mne.read_epochs(act_ep[0], verbose=False)
        act_pos_avgs.append(sub_active_epochs['1'].average())
        act_neg_avgs.append(sub_active_epochs['2'].average())
    except Exception as e:
        print(e)
        pass

pas_pos_avgs = []
pas_neg_avgs = []
pas_eps = [sorted(glob(deriv_dir+f'/sub-{sub_num:02d}_task-passive_run-all_event-stimtrack_epochs.fif') 
                for sub_num in passive_subs)]
for pas_ep in pas_eps[0]:
    try:
        sub_passive_epochs = mne.read_epochs(pas_ep[0], verbose=False)
        pas_pos_avgs.append(sub_passive_epochs['1'].average())
        pas_neg_avgs.append(sub_passive_epochs['2'].average())
    except:
        pass

# combien polarities
passive_avgs = [mne.combine_evoked([pas_pos_avgs[x], pas_neg_avgs[x]], weights='nave')
                for x in range(len(pas_pos_avgs))]
active_avgs = [mne.combine_evoked([act_pos_avgs[x], act_neg_avgs[x]], weights='nave')
                for x in range(len(act_pos_avgs))]
```

```python
active_times_by_subject_3 = {onset['sub_num']: onset['x'] for onset in rater3 if onset['task'] == 'active'}
passive_times_by_subject_3 = {onset['sub_num']: onset['x'] for onset in rater3 if onset['task'] == 'passive'}

common_subjects_active_3 = sorted([s for s in set(active_times_by_subject_3.keys())])
common_subjects_passive_3 = sorted([s for s in set(passive_times_by_subject_3.keys())])

active_times_by_subject_2 = {onset['sub_num']: onset['x'] for onset in rater2 if onset['task'] == 'active'}
passive_times_by_subject_2 = {onset['sub_num']: onset['x'] for onset in rater2 if onset['task'] == 'passive'}

active_times_by_subject_1 = {onset['sub_num']: onset['x'] for onset in rater1 if onset['task'] == 'active'}
passive_times_by_subject_1 = {onset['sub_num']: onset['x'] for onset in rater1 if onset['task'] == 'passive'}
```

```python
print(common_subjects_active_3)
print(common_subjects_passive_3)
```

```python
for i, sub in enumerate(common_subjects_active_3):
    evk_a = active_avgs[i].crop(-0.04, 0.2)
    rater1_onset = active_times_by_subject_1[sub]
    rater2_onset = active_times_by_subject_2[sub]
    rater3_onset = active_times_by_subject_3[sub]
    print(f"rater 1: {rater1_onset}", f"rater 2: {rater2_onset}", f"rater 3: {rater3_onset}")
    plt.plot(evk_a.times, evk_a.data[0])
    plt.axvline(rater1_onset, color='r', label='Rater 1')
    plt.axvline(rater2_onset, color='g', label='Rater 2')
    plt.axvline(rater3_onset, color='b', label='Tiebreaker')
    plt.title(f"Active FFR for sub-{sub:02d}")
    plt.legend()
    plt.show()
```

```python
for i, sub in enumerate(common_subjects_passive_3):
    evk_a = passive_avgs[i].crop(-0.04, 0.2)
    rater1_onset = passive_times_by_subject_1[sub]
    rater2_onset = passive_times_by_subject_2[sub]
    rater3_onset = passive_times_by_subject_3[sub]
    print(f"rater 1: {rater1_onset}", f"rater 2: {rater2_onset}", f"rater 3: {rater3_onset}")
    plt.plot(evk_a.times, evk_a.data[0])
    plt.axvline(rater1_onset, color='r', label='Rater 1')
    plt.axvline(rater2_onset, color='g', label='Rater 2')
    plt.axvline(rater3_onset, color='b', label='Tiebreaker')
    plt.title(f"Passive FFR for sub-{sub:02d}")
    plt.legend()
    plt.show()
```
